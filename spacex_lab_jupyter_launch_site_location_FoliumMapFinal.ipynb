Next, we need to explore and analyze the proximities of launch sites.


Let's first add a `MousePosition` on the map to get coordinate for a mouse over a point on the map. As such, while you are exploring the map, you can easily find the coordinates of any points of interests (such as railway)


```python
# Add Mouse Position to get the coordinate (Lat, Long) for a mouse over on the map
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)
site_map
```

Now zoom in to a launch site and explore its proximity to see if you can easily find any railway, highway, coastline, etc. Move your mouse to these points and mark down their coordinates (shown on the top-left) in order to the distance to the launch site.


```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

*TODO:* Mark down a point on the closest coastline using MousePosition and calculate the distance between the coastline point and the launch site.


```python
# find coordinate of the closet coastline
# e.g.,: Lat: 28.56367  Lon: -80.57163
launch_site_lat = 28.562302
launch_site_lon = -80.577356
coastline_lat = 28.56266
coastline_lon = -80.56822
distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
```

```python
# Create and add a folium.Marker on your selected closest coastline point on the map
# Display the distance between coastline point and launch site using the icon property
coastline_coordinate = [coastline_lat, coastline_lon]
distance_marker = folium.Marker(
    location=coastline_coordinate,
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html='<div style="font-size: 12px; color:#d35400;"><b>{:10.2f} KM</b></div>'.format(distance_coastline),
    )
)
site_map.add_child(distance_marker)
site_map
```

*TODO:* Draw a `PolyLine` between a launch site to the selected coastline point


```python
# Create a `folium.PolyLine` object using the coastline coordinates and launch site coordinate
coordinates = [[launch_site_lat, launch_site_lon], [coastline_lat, coastline_lon]]
lines = folium.PolyLine(locations=coordinates, weight=1)
site_map.add_child(lines)
site_map
```

```python
# Define launch site and target coordinates
launch_coord = [28.562302, -80.577356]
highway_coord = [28.56254, -80.57068]
railway_coord = [28.57205, -80.58525]
city_coord = [28.61200, -80.80788]

# Calculate distances
dist_highway = calculate_distance(launch_coord[0], launch_coord[1], highway_coord[0], highway_coord[1])
dist_railway = calculate_distance(launch_coord[0], launch_coord[1], railway_coord[0], railway_coord[1])
dist_city = calculate_distance(launch_coord[0], launch_coord[1], city_coord[0], city_coord[1])

proximities = [
    {"name": "Highway", "coord": highway_coord, "distance": dist_highway, "color": "#3498db"},
    {"name": "Railway", "coord": railway_coord, "distance": dist_railway, "color": "#2ecc71"},
    {"name": "City", "coord": city_coord, "distance": dist_city, "color": "#9b59b6"}
]

for prox in proximities:
    # Add a custom distance text marker
    folium.Marker(
        location=prox["coord"],
        icon=DivIcon(
            icon_size=(150, 36),
            icon_anchor=(0, 0),
            html=f'<div style="font-size: 12px; color:{prox["color"]};"><b>{prox["name"]}: {prox["distance"]:.2f} KM</b></div>'
        )
    ).add_child(folium.Popup(prox["name"])).add_to(site_map)

    # Draw a line between the launch site and target
    folium.PolyLine(
        locations=[launch_coord, prox["coord"]],
        weight=2.5,
        color=prox["color"]
    ).add_to(site_map)

site_map
```

After you plot distance lines to the proximities, you can answer the following questions easily:

*   Are launch sites in close proximity to railways?
Yes (approx. $1.33\text{ km}$$1.33\text{ km}$). Launch sites are kept very close to railway networks (like the NASA Railroad) because rockets, massive boosters, heavy payloads, and propellants require specialized heavy-rail transport to be delivered safely from manufacturing facilities to the assembly building and launchpad.

*   Are launch sites in close proximity to highways?
Yes (approx. $0.65\text{ km}$$0.65\text{ km}$). Close proximity to highways (such as the Samuel C. Phillips Parkway) is critical for transporting personnel, smaller hardware parts, and daily operational logistics to and from the site.

*   Are launch sites in close proximity to coastline?
Yes (approx. $0.89\text{ km}$$0.89\text{ km}$). This is perhaps the most critical flight safety requirement. Launching over the open ocean ensures that if an anomaly occurs during early flight, booster debris or falling payload fairings will safely land in the water rather than over populated areas.

*   Do launch sites keep certain distance away from cities?

Yes (approx. $23.18\text{ km}$$23.18\text{ km}$ to Titusville). Launch sites are intentionally built far away from major cities and heavily populated residential zones to protect communities from the extreme acoustic noise, shockwaves, and potential explosion hazards associated with rocket launches.
